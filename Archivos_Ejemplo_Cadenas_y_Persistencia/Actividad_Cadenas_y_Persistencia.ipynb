## 1. Diágnostico de cadenas

Antes de programar analicemos algunas de las operaciones para cadenas de texto en Python utilizando de base **codigo = &quot;  mx-gdl-0042  &quot;**:
- **len(codigo): `15`**. Es el total de caracteres almacenados en la variable codigo icluyendo los espacios extremos. Tiene 11 caracteres itnernos, y 2 espacios en cada extremo.
- **codigo[2:4]: `'mx'`**. Corta la cadena desde el índice 2 hasta el índice 4, sin tocar este último y recordando que en Python los índices se cuentan desde el 0.
- **codigo[-6:-2]: `'0042'`**. Los índices negativos cuentan desde el último caracter, corta la cadena desde el sexto caracter hasta el segundo caracter, ambos de izquierda a derecha, sin abarcar este último.
- **codigo.strip().upper(): `'MX-GDL-0042'`**. Remueve los espacios externos y convierte todo a mayúsculas.
- **codigo.find(&quot;xyz&quot;): `-1`**. Devuelve -1 porque no ecuentra la subcadena dentro del texto. Si sí encontrara el fragmento devolviera el índice de la posición inicial de la subcadena dentro del texto.

Verifiquemos las respuestas anteriores ejecutando el código en Python:

````python
codigo = "  mx-gdl-0042  "
print(f"len(codigo): {len(codigo)}")
print(f"codigo[2:4]: {codigo[2:4]}")
print(f"codigo[-6:-2]: {codigo[-6:-2]}")
print(f"codigo.strip().upper(): {codigo.strip().upper()}")
print(f"codigo.find('xyz'): {codigo.find('xyz')}")
````

Respondamos las siguientes preguntas:
- **¿codigo.strip() modifica codigo si no se reasigna?** No, en Python las cadenas son inmutables, es decir ninguna operación altera la cadena original almacenada en memoria. No modifica la variable si no que devuelve una cadena sin los espacios etremos. Si no se reasigna la variable "codigo" esta se mantendrá igual.
- **¿Qué diferencia existe entre in y find()?** El operador `in` devuelve un `True` o `False`, es útil si lo úncico que queremos saber es si lo contiene o no. Por otra parte, `find()` devuelve la posición del índice donde se encuentra, o un `-1` si no lo encuentra.
- **¿Por qué un resultado -1 no debe utilizarse como posición encontrada?** Porque el resultado de `-1` en `find()` es el resultado que arroja por default cuando no encuentra la subcadena dentro de la cadena principal. No está indicando la posición -1, la cual es el último caracter.

Analicemos el archivo "etiquetas.txt" contando cuántas categorías distintas tiene si sólo eliminamos filas vacías, y comparándolo con cuántas categorías hay después normalizar el texto. 


````python
from pathlib import Path 

BASE = Path('.').resolve()
etiquetas = BASE / 'datos' / 'etiquetas.txt' 

with etiquetas.open('r', encoding='utf-8') as e: 
    # Quitamos filas vacías y eliminamos saltos de línea al final de cada fila
    filas = [linea.rstrip('\r\n') for linea in e if linea.strip()]

categorias_og= set(filas) 
print( 'Categorías únicas antes de normalizar:', len(categorias_og), ) 

categorias_limpias = { '_'.join( #unir palabras con guión bajo
    fila.strip() #quitamos espacios al inicio y final
    .lower() #convertimos a minúsculas
    .replace('-', ' ') #reemplazamos guiones por espacios
    .replace('_', ' ') #reemplazamos guiones bajos por espacios
    .split() ) #separamos palabras
    for fila in filas } 
print('Categorías únicas después de normalizar:', len(categorias_limpias))
````

### 2.2. Interpretación de resultados de `ejemplo_guiado.py`

El archivo `errores.txt` se abre en modo "w" o *write* (escritura): `(SALIDA / "errores.txt").open("w", encoding="utf-8")`. Esto hace que el archivo se abra con el fin de escribirlo desde cero, si no existía el archivo lo crea desde cero, pero si el archivo ya existía elimina su contenido y vuelve a escribirlo desde cero. Es por esto que ste archivo sólo conserva dos líne, sin importar las veces que lo ejecutes.
Por otra parte, el archivo `ejecuciones.log` se abre en modo "a" o *append* (anexado): `(SALIDA / "ejecuciones.log").open("a", encoding="utf-8")`. Si el archivo no existía lo crea, pero si el archivo tenía registros previos los conserva y añade el nuevo texto al final. Cada vez que se corre el programa se agrega una línea nueva al final del archivo, es por esto que sigue aumentantando, a diferencia del achivo anterior.

En el ejemplo guiado se utilizan varias operaciones, algunos ejemplosson los siguientes:
- `strip()`: se usa en `etiqueta.strip().split(":")`. Su función es eliminar los espacios en blanco en los extremos de la cadena " sensor\_norte:23.5 ".
- `split()`: se usa en `etiqueta.strip().split(":")`. Divide la cadena usando el caracter ":" como separador. Entonces la separa en dos variables: `nombre` (`"sensor_norte"`) y `valor` (`"23.5"`).
- `replace()`: se usa en `nombre.replace("_", " ")`. Reemplaza el guión bajo "_" por un espacio, transformando `"sensor_norte"` en `"sensor norte"`.
- `capitalize()`: se usa en 
**capitalize()**: `print(nombre.replace("_", " ").capitalize())`. Convierte la primera letra de la cadena en mayúscula y el resto en mínusculas. Como resultado se imprime `"Sensor norte"`.
- `f-strings`: se usa en  `print(f"Valor: {float(valor):.1f}")`, en `print(f"Errores detectados: {total}\n"` y `f"Errores: {total}")` y en `print(f"Errores: {total})"`. Se usa para imprimir texto impreso junto con variables en la bitácora. Cuando se utiliza `.f` como en `float(valor):.1f`, esto formatea el número flotante a que sólo tenga cierta cantidad de decimales, en el ejemplo sólo tendría uno.
- **Cierre automático de archivos**: se implementa con el uso de **with** (`with (BASE / "datos" / "eventos.txt").open(...) as entrada:`). Se usa para asegurarnos de que los archivos se cierren  inmediatamente después de ejecutar ese bloque de código, sin importar si ocurre un error. ## 3. Implementación del depurador

En esta sección buscamos construir las funciones para limpiar, validar y filtrar los datos de `solicitudes.txt` usando como base el archivo `actividad_inicio.py`.


### 3.1. Normalización de categorías

La función `normalizar_categoria` estandariza variaciones las variaciones en cómo está escrita la categoría para unificarlas y n tener duplicados por variaciones en el formato.

````python
def normalizar_categoria(texto): 
    """Normaliza la categoría de una solicitud: 
    1. Elimina espacios extremos (strip) y pasa a minúsculas (lower). 
    2. Sustituye guiones '-' y guiones bajos '_' por espacios. 
    3. Separa palabras y las une con guion bajo '_', conservando tildes.  """
    # strip(): quitar espacios, lower(): pasar a minúsculas
    # replace("-", " "): reemplazar guiones, replace("_", " "): remplazar guiones bajos por espacios
    texto_limpio = texto.strip().lower().replace("-", " ").replace("_", " ") 
    palabras = texto_limpio.split()  #separar palabras
    return "_".join(palabras) #unir palabras con guion bajo
````

### 3.2. Validación de campos

La función `validar_campos` evalúa cada línea dividida por el separador `|`. Recibe la lista `campos` y el conjunto `ids_vistos`. Se validarán los campos siguendo el siguiente orden de evaluación obligatorio:
1. Línea vacía
2. Cantidad de campos distinta de cuatro
3. Algún campo vacío tras `strip()`
4. ID con formato inválido
5. Correo inválido
6. Categoría fuera del catálogo
7. ID duplicado
8. Limpiar descripción

Un ID debe agregarse al conjunto de IDs vistos únicamente después de que el registro cumple todas las validaciones anteriores y es aceptado por completo.

````python
def validar_campos(campos, ids_vistos): 
    """Evalúa un registro según el orden estricto de reglas de calidad. 
    Regresa (registro_limpio, motivo). 
    - Rechazo: (None, motivo) 
    - Válido: (dict_limpio, None) """ 
    # 1. Línea vacía 
    if len(campos) == 1 and not campos[0].strip(): # Comprueba si la lista está vacía o si el único elemento es una cadena vacía
        return None, "línea vacía" 
    
    # 2. Cantidad de campos distinta de cuatro 
    if len(campos) != 4: # Comprueba si la lista tiene exactamente cuatro elementos
        return None, "cantidad de campos distinta de cuatro" 
    
    campos_strip = [c.strip() for c in campos] # Elimina espacios al inicio y final de cada campo

    # 3. Campo vacío tras strip() 
    if any(len(c) == 0 for c in campos_strip): # Comprueba si algún campo está vacío después de eliminar espacios con strip()
        return None, "algún campo vacío tras strip()" 
    
    id_og, correo_og, cat_og, desc_og = campos_strip  # Guardar los campos en distintas variables

    # 4. ID con formato inválido
    id_norm = id_og.upper() # Convertir a mayúsculas
    if not (len(id_norm) == 8  # Comprueba si la longitud del ID es 8
            and id_norm.startswith("SOL-") # Se asegura de que el ID comience con "SOL-"
            and id_norm[4:].isdigit() # Comprueba si los últimos 4 caracteres del ID son números enteros del 0 al 9
            ): 
         return None, "ID con formato inválido" 
    
    # 5. Correo inválido 
    correo_norm = correo_og.lower() # Convertir a minúsculas
    if " " in correo_norm or correo_norm.count("@") != 1: # Comprueba si tiene espacios o si tiene sólo un "@"
        return None, "correo inválido" 
    
    usuario, dominio = correo_norm.split("@")  # Separa el correo en usuario y dominio
    if ( not usuario or not dominio # Comprueba si el usuario o el dominio están vacíos
        or "." not in dominio  # Comprueba si el dominio no tiene un punto decimal
        or dominio.startswith(".") or dominio.endswith(".") ): # Comprueba si el dominio comienza o termina con un punto decimal
        return None, "correo inválido" 
    
    # 6. Categoría fuera del catálogo 
    cat_norm = normalizar_categoria(cat_og) # Aplicar la función de normalización de categoría
    cat_catalogo = {"acceso", "pago_en_línea", "soporte_técnico"} # Definir el catálogo de categorías válidas
    if cat_norm not in cat_catalogo: # Comprueba si la categoría normalizada no está en el catálogo
        return None, "categoría fuera del catálogo" 
    
    # 7. ID duplicado 
    if id_norm in ids_vistos: # Comprueba si el ID normalizado ya se guardó en IDs vistos
        return None, "ID duplicado" 
    
    # 8. Limpieza de descripción 
    desc_norm = " ".join(desc_og.split())  # split() separa las palabras por espacios (elimina espacios extra)
    #join() une las palabras separadas con un espacio para volver a fomar la cadena
    if not desc_norm: 
        return None, "algún campo vacío tras strip()" 
    registro_limpio = { "id": id_norm, 
                       "correo": correo_norm, 
                       "categoria": cat_norm, 
                       "descripcion": desc_norm, }  # Creamos un diccionario con los campos limpios
    return registro_limpio, None
````

## 4. Persistencia y verficación
### 4.1. Persitencia de datos
En esta fase se crea la función `main()`, en esta se unifica todo el flujo anterior. Primero debemos abrir el archivo con `encoding="utf-8"` y verificar que la primera línea coincida exactamente con `id|correo|categoria|descripcion`.

````python
def main(): 
    try: 
        # Crear carpeta de salida si no existe 
        SALIDA.mkdir(parents=True, exist_ok=True) 
        
        # Inicializar variables
        ids_vistos = set() 
        total_lineas = 0 
        validos_cnt = 0 
        rechazados_cnt = 0 
        por_categoria = {"acceso": 0, "pago_en_línea": 0, "soporte_técnico": 0} 
        registros_limpios = [] 
        lista_rechazos = [] 

        # Abrir archivo de entrada de forma segura con with
        with ENTRADA.open("r", encoding="utf-8") as archivo: 

            encabezado = archivo.readline() # Leer la primera línea del archivo como encabezado
            if not encabezado: # Comprobar si el encabezado está vacío
                raise ValueError("El archivo de entrada está vacío.") 
            
            encabezado_clean = encabezado.strip() # Eliminar espacios al inicio y final del encabezado
            encabezado_esperado = "id|correo|categoria|descripcion" 
            if encabezado_clean != encabezado_esperado: # Comprobar si el encabezado coincide con el esperado
                raise ValueError( 
                    f"Encabezado incorrecto. Se esperaba '{encabezado_esperado}', pero" 
                    f" se recibió '{encabezado_clean}'." ) 
            
            # Procesar cada línea del archivo a partir de la segunda línea
            for num_linea, linea in enumerate(archivo, start=2): 
                total_lineas += 1 
                linea_raw = linea.rstrip("\r\n") # Eliminar saltos de línea al final de la línea
                campos = linea_raw.split("|") # Separar la línea en campos usando "|" como delimitador

                # Guardar el registro limpio o el motivo de rechazo según el resultado de la validación
                registro, motivo = validar_campos(campos, ids_vistos) 

                # Si el registro es válido, agregarlo a la lista de registros limpios y actualizar los contadores y categorías
                if registro is not None:
                    ids_vistos.add(registro["id"]) 
                    validos_cnt += 1 
                    por_categoria[registro["categoria"]] += 1
                    registros_limpios.append(registro) 
                # Si el registro es rechazado, agregar el motivo de rechazo a la lista de rechazos y actualizar el contador de rechazados    
                else: 
                    rechazados_cnt += 1 
                    lista_rechazos.append(
                        { "linea": num_linea, 
                         "motivo": motivo, 
                         "texto_original": linea_raw, })

        
        # Archivos de salida
        # 1. solicitudes_limpias.txt
        archivo_limpias = SALIDA / "solicitudes_limpias.txt" 
        with archivo_limpias.open("w", encoding="utf-8") as f_limpias: 
            f_limpias.write("id|correo|categoria|descripcion\n") 
            for reg in registros_limpios: 
                f_limpias.write( f"{reg['id']}|{reg['correo']}|{reg['categoria']}|{reg['descripcion']}\n" )
        
        # 2. rechazos.jsonl
        archivo_rechazos = SALIDA / "rechazos.jsonl"
        with archivo_rechazos.open("w", encoding="utf-8") as f_rechazos: 
            for rechazo in lista_rechazos: 
                f_rechazos.write(json.dumps(rechazo, ensure_ascii=False) + "\n")

        # 3. resumen.json
        resumen_data = { 
            "total": total_lineas, 
            "validos": validos_cnt, 
            "rechazados": rechazados_cnt, 
            "por_categoria": por_categoria, } 
        archivo_resumen = SALIDA / "resumen.json" 
        with archivo_resumen.open("w", encoding="utf-8") as f_resumen: 
            json.dump(resumen_data, f_resumen, ensure_ascii=False, indent=2)

        # 4. ejecuciones.log
        archivo_log = SALIDA / "ejecuciones.log" 
        with archivo_log.open("a", encoding="utf-8") as f_log: 
            f_log.write( f"[{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] Entrada:" 
                        f" {ENTRADA.name} | Total: {total_lineas} | Válidos: {validos_cnt} |" 
                        f" Rechazados: {rechazados_cnt}\\n" )


        # Verificacion de resultados
        # Contar registros limpios exclyendo encabezado
        with archivo_limpias.open("r", encoding="utf-8") as f_chk_limpias: 
            cnt_limpias_guardadas = len(f_chk_limpias.readlines()) - 1

        # Valida rechazos.jsonl
        cnt_rechazos_guardados = 0
        with archivo_rechazos.open("r", encoding="utf-8") as f_chk_rechazos:
            for l in f_chk_rechazos:
                if l.strip():
                    json.loads(l)
                cnt_rechazos_guardados += 1

        # Cargar resumen.json 
        with archivo_resumen.open("r", encoding="utf-8") as f_chk_resumen: 
            resumen_cargado = json.load(f_chk_resumen)

        # Verificar conteos
        assert (cnt_limpias_guardadas == validos_cnt ), "Error en solicitudes_limpias.txt" 
        assert (cnt_rechazos_guardados == rechazados_cnt ), "Error en rechazos.jsonl" 
        assert ( resumen_cargado["total"] == total_lineas ), "Error en resumen.json" 
        assert ( resumen_cargado["validos"] + resumen_cargado["rechazados"] == total_lineas ), "La suma no coincide" 
        print( f"Total procesados: {total_lineas} | Válidos: {validos_cnt} |" f" Rechazados: {rechazados_cnt}" ) 
        print(f"Por categoría: {por_categoria}")
     
    except FileNotFoundError as error: 
        print(f"Error: No se encontró el archivo -&gt; {error.filename}") 
    except ValueError as error: 
        print(f"Error estructural en la entrada: {error}")
````
